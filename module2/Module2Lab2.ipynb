This notebook demonstrates how multiplication can be performed through repeated addition rather than the multiplication operator. It includes two implementations: one using a for loop and another using a while loop.

```python
def multiply_with_for(number, repetitions):
    result = 0

    for count in range(repetitions):
        result = result + number

    return result
```

```python
def multiply_with_while(number, repetitions):
    result = 0
    count = 0

    while count < repetitions:
        result = result + number
        count = count + 1

    return result
```

```python
first_number = 7
second_number = 6

for_result = multiply_with_for(first_number, second_number)
while_result = multiply_with_while(first_number, second_number)
real_product = first_number * second_number

print("For-loop result:", for_result)
print("While-loop result:", while_result)
print("Python product:", real_product)
print("Both functions agree:", for_result == while_result == real_product)
```

## Conclusion

Both functions correctly returned 42 for {$7 \times 6$}. The for-loop and while-loop implementations achieve multiplication by repeatedly adding 7 a total of six times. This illustrates how loops can accumulate a running total and ultimately return the final result.
