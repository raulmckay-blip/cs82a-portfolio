# Problem 4.4: Temperature Categories

This notebook uses a loop together with an `if`/`elif`/`else` chain to assign a category to each temperature reading.

**Temperature cutoffs:**
- **Cold:** below 15°C
- **Warm:** 15°C through 25°C
- **Hot:** above 25°C

```python
temperatures = [10, 16, 23, 28, 14, 22, 30]

for temperature in temperatures:
    if temperature < 15:
        category = "cold"
    elif temperature <= 25:
        category = "warm"
    else:
        category = "hot"

    print(f"{temperature}°C is {category}.")
```

## Conclusion

The loop successfully classified every temperature in the list. Temperatures below 15°C were labeled cold, those from 15°C to 25°C were labeled warm, and anything above 25°C was labeled hot. This shows how a loop paired with a conditional statements can efficiently categorize data.
