```python
temps = [61, 65, 72, 58, 70, 74, 68]
warm_days = 0

for temperature in temps:
    if temperature > 65:
        warm_days = warm_days + 1

print(warm_days)
```

```python
def to_celsius(f):
    return (f - 32) * 5 / 9

converted_temperature = to_celsius(temps[0])

print(round(converted_temperature, 1))
```
