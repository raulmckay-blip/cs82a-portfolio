## Overview

This is a one-page note on what it would take to move my course work from Google Colab to GitHub Codespaces. I used Colab for Lab 1 and for a long project I had before this class, so most of this comes from that.

## Where I'm moving from and to

- **Now:** Google Colab (my notebooks live in my Google Drive)
- **Moving to:** GitHub Codespaces, which is basically VS Code running in the cloud, opened right from my GitHub repo. I picked it because I already use VS Code and GitHub for every lab in this class, so it feels like the most natural move for me.

## What transfers untouched

- **The notebook itself.** A `.ipynb` is a `.ipynb`. It's already in my `cs82a-portfolio` repo, so it just shows up in the codespace.
- **The Python code.** Code like my `convert_lbs_to_kgs` function from Lab 1 runs the same no matter where it is.
- **The data files and paths.** My datasets already sit in each module folder next to the notebooks, so `pd.read_csv("penguins.csv")` works the same as on my laptop. No extra upload step like in Colab, which is definitely a win.
- **The editor.** It's literally VS Code in the browser, same code cells, same markdown cells, same shortcuts.

## What needs redoing

- **Accounts / sign-in:** Nothing new, I just sign in with the GitHub account I already push my labs to.
- **File storage / datasets:** Anything in my `.gitignore` won't be there, since the codespace only gets what's on GitHub. For example, my Module 6 database and customer CSV are kept off GitHub on purpose, so I'd have to bring those in another way.
- **Package installs:** This one is a step back from Colab. The libraries aren't all set up for me, so I'd have to run `pip install pandas matplotlib seaborn` (and add the Jupyter extension if it asks) every time I make a new codespace.
- **Sharing:** In Colab sharing is as simple as sharing a Google Doc, just a link. With Codespaces the other person needs access to my repo, so quick sharing is harder.

## What I actually ran into (local VS Code → Colab)

- **No installs.** I didn't have to install Python or any libraries, which is definitely a cool part. If I'm working on a new device, this is a win for sure.
- **Signing in** was as simple as signing in to Gmail or a Google Doc, just an email and a password.
- **The setup feels familiar.** Same code cells and markdown cells as VS Code.
- **Uploading a CSV is an extra step.** In VS Code the file just sits in my folder next to the notebook. In Colab I literally have to upload the file to the session first.
- **Sharing** is as simple as sharing a Google Doc or a Google Sheet. A link with the right access is enough for the other person to open it and work on the same project. That seems like a massive benefit when working with someone, and I guess that's why Google named it Colab, which makes sense.
- **The order of running cells feels more important than in VS Code.** If the runtime restarts, nothing I ran before is still there, so I have to run everything again from the top. For Lab 1 I used *Restart session and run all* before saving so the cells came out clean as [1], [2], [3].

## One risk

**Risk:** Personal info ending up in the cloud, or worse, on GitHub. Say a notebook needs a CSV with info I don't want online. To use it in the codespace I'd have to upload it there, so even if I never commit it, the file is now sitting on GitHub's servers instead of my laptop. And if it's not in my `.gitignore`, one `git add .` and it's on my public repo for anyone to see.

I actually had a close call with this. In Module 6, `store_customers.csv` had customer names and cities. An earlier module had already taught us to be careful about what we upload or commit, so before pushing I stopped, checked what personal info was in the file, and added it to `.gitignore`. If I hadn't, a normal push would have put those names out there, and with real customer data that would have been a privacy breach.

This isn't just a Codespaces thing either. Uploading a CSV to Colab also puts it on Google's servers. Any cloud platform means trusting someone else with your data.

**How I'd handle it:**

- Take out the personal columns (names, emails, phone numbers) before uploading, or just use fake sample data
- Add those files to `.gitignore` *before* the first commit
- Delete them from the codespace when I'm done
- Keep anything really sensitive on my own laptop
