## Problem 6.1: Pounds to kilograms

1 lb = 0.453592 kg

```python
def convert_lbs_to_kgs(lbs):
    return lbs * 0.453592
```

## Testing it on three values

One of them is my own weight.

```python
# Test 1: My own weight
print("My weight in kg:", round(convert_lbs_to_kgs(180),2))

# Test 2: A lighter value
print("1 lbs in kg:", round(convert_lbs_to_kgs(1),2))

# Test 3: A heavier value
print("125,000 lbs in kg:", round(convert_lbs_to_kgs(125000),2))
```
