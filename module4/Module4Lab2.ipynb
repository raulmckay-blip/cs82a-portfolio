## Overview

This lab uses `allbacks.csv`, which has 15 books with their `volume` (cm³), cover `area` (cm², 0 for paperbacks), `weight` (g), and `cover` type (`hb` = hardcover, `pb` = paperback). The goal is to find the median cover type, the mean weight of all the books, and the mean weight for each cover type, then compare the two groups.

## Load the data

```python
import pandas as pd

books = pd.read_csv('allbacks.csv', index_col=0)
print(books.shape)   # expect (15, 4)
books
```

```python
books['cover'].value_counts()
```

## Median cover type

Cover type isn't a number, but it still has an order. Paperback is the cheaper, lower binding and hardcover is the higher one. So I set the order as `pb` < `hb`, sort all the books, and take the middle one, same as a median with numbers.

```python
cover_order = pd.CategoricalDtype(['pb', 'hb'], ordered=True)
covers = books['cover'].astype(cover_order).sort_values().reset_index(drop=True)

n = len(covers)
middle = n // 2          # index 7 = the 8th book
median_cover = covers[middle]

print(list(covers))
print(f'n = {n}, middle position = {middle + 1}, median cover = {median_cover}')
```

Note: The median cover type is paperback. 

## Mean weight of all books

```python
mean_weight = books['weight'].mean()
print(f'Mean weight of all books: {mean_weight:.2f} g')
```

Note: The average book weighs about 707 g, which is a little over 1.5 lb.

## Mean weight by cover type

```python
group_means = books.groupby('cover')['weight'].mean()
print(group_means.round(2))

diff = group_means['hb'] - group_means['pb']
print(f'\nHardcovers are {diff:.2f} g heavier on average')
```

Note; Hardcovers average about 796 g and paperbacks about 628 g, so hardcovers are around 168 g heavier.

## Checking book size

Before saying the cover is what makes the difference, I wanted to check if the hardcovers are just bigger books. If they were, that could explain the extra weight instead of the cover.

```python
books.groupby('cover')['volume'].mean().round(1)
```

## What the difference suggests

The two groups are pretty much the same size:

Paperbacks are actually a little bigger on average (856 vs 834 cm³). Even so, hardcovers still weigh about 168 g more, so it looks like the hard cover itself adds real weight.
